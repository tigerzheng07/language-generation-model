```python
!pip install transformers torch

from transformers import pipeline, set_seed

generator = pipeline("text-generation", model="gpt2")
set_seed(42)
print("===简单文本生成示例===")
prompt = "artifitial intelligence can"
results = generator(prompt, max_length = 50, num_return_sequences = 1)
print(results[0]["generated_text"])
```

```python
print("===互动文本生成===")
def interactive_advanced_generator():
  while True:
    prompt = input("输入一个文本开头,输入quit退出")
    if prompt.lower() == "quit":
      break
    length = input("请输入文本长度(默认100)")
    length = int(length) if length.isdigit() else 100

    num_samples = input("请输入样本的生成数(1-3,默认1)")
    num_samples = int(num_samples) if num_samples.isdigit() and 1 <= int(num_samples) <= 3 else 1
    print("\n生成中...")
    results = generator(prompt, max_length = length, num_return_sequences = num_samples, temperature = 0.7, do_sample = True)
    for i, result in enumerate(results):
      print(f"\n样本 {i + 1}")
      print(result["generated_text"])

interactive_advanced_generator()
```
